### Activity 01

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv")
df.head()
```

```python
df.describe()
```

```python
df["bill_length_mm"].plot(kind = "line")
plt.title("Line plot bill length (mm)")
plt.grid()
plt.show()
```

```python
plt.scatter(x = df[df.species == "Adelie"]["body_mass_g"],
            y = df[df.species == "Adelie"]["bill_length_mm"])

plt.scatter(x = df[df.species == "Gentoo"]["body_mass_g"],
            y = df[df.species == "Gentoo"]["bill_length_mm"])

plt.scatter(x = df[df.species == "Chinstrap"]["body_mass_g"],
            y = df[df.species == "Chinstrap"]["bill_length_mm"])
plt.grid()
plt.title("Scatter plot")
plt.xlabel("Body Mass (g)")
plt.ylabel("Bill Length (mm)")
plt.show()
```
