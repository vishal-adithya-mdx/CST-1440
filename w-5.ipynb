## Activity 08

```python
# A = The integer picked is divisible by 3
# B = The integer picked is divisible by 4
# C = The integer picked is divisible by 6

S = set(np.arange(1,25))
A = set(np.arange(3,25,3))
B = set(np.arange(4,25,4))
C = set(np.arange(6,25,6))

A_B = A.intersection(B)
A_B_ = A.union(B)

B_C_ = C.intersection(B)
A_C = A.intersection(C)

B_C = len(B.intersection(C)) / len(C)
C_B = len(B.intersection(C)) / len(B)

print(f"P(A):   {len(A)/len(S):10.2f}")
print(f"P(B):   {len(B)/len(S):10.2f}")
print(f"P(AvB): {len(A_B_)/len(S):10.2f}")
print(f"P(A∧B): {len(A_B)/len(S):10.2f}")
print(f"P(B/C): {B_C:10.2f}")
print(f"P(C/B): {C_B:10.2f}")
```

```python
print(f"Independence of Events: A & B - {len(A_B)/len(S) == (len(A)/len(S)) * (len(B)/len(S))} ")
print(f"Independence of Events: B & C - {len(B_C_)/len(S) == (len(C)/len(S)) * (len(B)/len(S))} ")
print(f"Independence of Events: A & C - {len(A_C)/len(S) == (len(A)/len(S)) * (len(C)/len(S))} ")
```

```python
S = 0.2
Sc = 1 - S

L_S = 0.7 # = L n S / S
L_Sc = 0.05

U_S = 0.6
U_Sc = 0.1
```

```python
# L n U = LU_S + LU_Sc  
a1 = L_S * U_S
a2 = L_Sc * U_Sc


LU = ((a1) * S) + ((a2)* Sc) 
a3 = ((a1) * S) / LU
```
