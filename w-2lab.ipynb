### Lists

```python
fruit = ['apple', 'banana', 'cherry', 'durian', 'elderberry']

print(f"Element at index 1: {fruit[1]}")
print(f"Element at index -2: {fruit[-2]}")
print(f"First two elements: {fruit[:2]}")
print(f"All elements starting fom index 3: {fruit[3:]}")
print(f"Elements from index 1 to 4: {fruit[1:4]}")
```

```python
fruit.append("blueberry")
print(fruit)

# the append method adds a new element in list. 
```

```python
fruit.count("banana")

# the count method returns the no of times the element has occured in the list.
```

```python
print(fruit)
fruit.pop()
print(fruit)
```

### Importing Dataset in Pandas

#### Step 1

```python
flights = pd.read_csv("https://vincentarelbundock.github.io/Rdatasets/csv/nycflights13/flights.csv")
flights.head()
```

#### Step 2

```python
flights.describe()
```

```python
flights.shape
```

```python
flights.dtypes
```

```python
flights.isnull().sum()
```

```python
flights.dropna(how = "any",inplace = True)
flights.isnull().sum()
```

#### Q1.

```python
q1 = flights[flights["dep_delay"] >=120]
q1.head()
```

#### Q2.

```python
q2 = flights[flights["dest"] == "IAH"]
q2_ = flights[flights["dest"] == "HOU"]
q2 = pd.concat([q2,q2_])
q2.head()
```

#### Q3.

```python
q3 = flights[flights["arr_delay"] > 120]
q3 = q3[q3["dep_delay"] <= 0]
q3.head()
```

#### Q4.

```python
flights["time_gained"] = flights["dep_delay"] - flights["arr_delay"]

q4 = flights[flights["dep_delay"]>=60]
q4 = q4[q4["time_gained"] > 30]
q4.head()
```

#### Q5.

```python
airtime = flights[flights["air_time"] > 0]
flights["avg_mph"] = flights["distance"] / airtime["air_time"]
flights.head()
```

```python
q5 = flights.sort_values(by = ["avg_mph"],ascending=False)
q5.head()
```

#### Q6.

```python
q6 = flights.sort_values(by = ["time_gained"],ascending=False)
q6.head()
```
